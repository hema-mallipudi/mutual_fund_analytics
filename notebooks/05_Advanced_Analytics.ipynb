```python
#Task 1 - historical Var and CVar
nav['date'] = pd.to_datetime(nav['date'])

nav = nav.sort_values(['amfi_code','date'])

nav['daily_return'] = nav.groupby('amfi_code')['nav'].pct_change()
```

```python
risk_metrics = []

for scheme, group in nav.groupby('amfi_code'):
    
    returns = group['daily_return'].dropna()
    
    if len(returns) > 50:
        
        var95 = np.percentile(returns,5)
        
        cvar = returns[returns <= var95].mean()
        
        risk_metrics.append(
            [scheme,var95,cvar]
        )

risk_df = pd.DataFrame(
    risk_metrics,
    columns=['amfi_code','VaR95','CVaR']
)
print(risk_df.head())
```

#Task 2 — Rolling 90 Day Sharpe

```python
nav['rolling_sharpe'] = (
    nav.groupby('amfi_code')['daily_return']
    .transform(
        lambda x:
        x.rolling(90).mean()
        /
        x.rolling(90).std()
        *
        np.sqrt(252)
    )
)
```

```python
top5 = nav['amfi_code'].unique()[:5]
plt.figure(figsize=(12,6))
for fund in top5:
    temp = nav[
        nav['amfi_code']==fund
    ]
    plt.plot(
        temp['date'],
        temp['rolling_sharpe'],
        label=fund
    )
plt.legend()
plt.title(
    "Rolling 90 Day Sharpe"
)
plt.savefig(
    "rolling_sharpe_chart.png"
)
plt.show()
```

```python
# Step 1: Create Cohort Year- Purpose: Find the first transaction year for each investor.
transactions['transaction_date'] = pd.to_datetime(
    transactions['transaction_date']
)
cohort = (
    transactions
    .groupby('investor_id')['transaction_date']
    .min()
    .dt.year
    .reset_index()
)
cohort.columns = ['investor_id', 'cohort_year']
cohort.head()
```

```python
# Step 2: Merge Cohort Year into Transactions -Purpose: Add cohort year back to every transaction.
transactions = transactions.merge(
    cohort,
    on='investor_id',
    how='left'
)
transactions.head()
```

```python
# Step 3: Merge Scheme Name(Add Schema Name) - Purpose: Bring scheme names from Dataset 7.
transactions = transactions.merge(
    performance[['amfi_code', 'scheme_name']],
    on='amfi_code',
    how='left'
)
transactions.head()
```

```python
# Step 4: Clean Duplicate Columns - Purpose: Remove duplicate columns if merges were run multiple times.
transactions = transactions.rename(
    columns={
        'cohort_year_x': 'cohort_year',
        'scheme_name_x': 'scheme_name'
    }
)
transactions = transactions.drop(
    columns=['cohort_year_y', 'scheme_name_y'],
    errors='ignore'
)
transactions.columns
```

```python
#Average SIP Amount & Total invested
# Step 5: Cohort Summary - Purpose: Calculate average SIP amount and total invested per cohort.
cohort_summary = (
    transactions
    .groupby('cohort_year')
    .agg(
        avg_sip=('amount_inr', 'mean'),
        total_invested=('amount_inr', 'sum')
    )
    .reset_index()
)
cohort_summary
```

```python
#top fund preference per cohort
# Step 6: Top Fund Preference -Purpose: Find the most popular fund in each cohort.
top_fund = (
    transactions
    .groupby(['cohort_year', 'scheme_name'])
    .size()
    .reset_index(name='count')
)
top_fund = (
    top_fund
    .sort_values(
        ['cohort_year', 'count'],
        ascending=[True, False]
    )
    .groupby('cohort_year')
    .first()
    .reset_index()
)
top_fund = top_fund[
    ['cohort_year', 'scheme_name']
]
top_fund
```

```python
#Final Investor Cohort Analysis
# Step 7: Final Cohort Analysis -Purpose: Combine everything into one table.

cohort_analysis = cohort_summary.merge(
    top_fund,
    on='cohort_year',
    how='left'
)
cohort_analysis
```

```python
# Step 8: Save Result - Purpose: Export the result.
cohort_analysis.to_csv(
    'investor_cohort_analysis.csv',
    index=False
)
print("Investor Cohort Analysis Saved")
```

TASK 4 - SIP CONTINUITY ANALYSIS

```python
# Step 1: Keep only SIP transactions(filter SIP Transactions)
sip_txns = transactions[
    transactions['transaction_type'] == 'SIP'
].copy()
sip_txns.head()
```

```python
# Step 2: Sort transactions by Investor and Date
sip_txns['transaction_date'] = pd.to_datetime(
    sip_txns['transaction_date']
)
sip_txns = sip_txns.sort_values(
    ['investor_id', 'transaction_date']
)
sip_txns.head()
```

```python
# Step 3: Find investors with 6+ SIP transactions
sip_counts = (
    sip_txns
    .groupby('investor_id')
    .size()
)
eligible_investors = sip_counts[
    sip_counts >= 6
].index
sip_txns = sip_txns[
    sip_txns['investor_id'].isin(eligible_investors)
]
print("Eligible Investors:", len(eligible_investors))
```

```python
# Step 4: Calculate day gaps between consecutive SIPs
sip_txns['gap_days'] = (
    sip_txns
    .groupby('investor_id')['transaction_date']
    .diff()
    .dt.days
)
sip_txns.head()
```

```python
# Step 5: Average gap Per Investor
avg_gap = (
    sip_txns
    .groupby('investor_id')['gap_days']
    .mean()
    .reset_index()
)
avg_gap.columns = [
    'investor_id',
    'avg_gap_days'
]
avg_gap.head()
```

```python
# Step 6: Risk classification(Create Risk Flag)
avg_gap['status'] = avg_gap[
    'avg_gap_days'
].apply(
    lambda x: 'At-Risk'
    if x > 35
    else 'Active'
)
avg_gap.head()
```

```python
# Step 7: Final SIP Continuity report
sip_continuity_report = avg_gap.sort_values(
    'avg_gap_days',
    ascending=False
)
sip_continuity_report
```

```python
# Step 8: Summary
total_investors = len(sip_continuity_report)
at_risk = (
    sip_continuity_report['status']
    == 'At-Risk'
).sum()
active = (
    sip_continuity_report['status']
    == 'Active'
).sum()
print("Total Investors :", total_investors)
print("Active Investors:", active)
print("At-Risk Investors:", at_risk)

print(
    "Continuity Rate:",
    round(active / total_investors * 100, 2),
    "%"
)
```

```python
# Step 9: Save report
sip_continuity_report.to_csv(
    'sip_continuity_report.csv',
    index=False
)
print("Report Saved")
```

## Sector HHI Concentration Analysis

The Herfindahl-Hirschman Index (HHI) was used to measure portfolio concentration.

HHI = Σ(weight_pct²)

- Higher HHI indicates a concentrated portfolio.
- Lower HHI indicates a diversified portfolio.
- Funds with high HHI may carry higher stock-specific risk.

```python
# TASK 6 - Sector HHI Concentration

# Total AUM of all funds
total_aum = performance['aum_crore'].sum()

# Weight of each fund
performance['weight'] = performance['aum_crore'] / total_aum

# HHI = sum(weight²)
performance['weight_sq'] = performance['weight'] ** 2

hhi = (
    performance.groupby(['amfi_code', 'scheme_name'])['weight_sq']
    .sum()
    .reset_index()
)
hhi.columns = ['amfi_code', 'scheme_name', 'HHI']
hhi.head()
```

```python
#Compare across equity funds
equity_categories = [
    'Large Cap',
    'Mid Cap',
    'Small Cap',
    'Flexi Cap',
    'Large & Mid Cap'
]
equity_hhi = hhi.merge(
    performance[['amfi_code', 'category']],
    on='amfi_code',
    how='left'
)
equity_hhi = equity_hhi[
    equity_hhi['category'].isin(equity_categories)
]
equity_hhi.head()
```

```python
#Top 10 most concentated funds
most_concentrated = (
    equity_hhi
    .sort_values('HHI', ascending=False)
    .head(10)
)

most_concentrated
```

```python
#top  most diversified funds
most_diversified = (
    equity_hhi
    .sort_values('HHI', ascending=True)
    .head(10)
)

most_diversified
```
